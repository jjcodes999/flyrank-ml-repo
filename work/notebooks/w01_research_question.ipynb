## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Answer:

I looked at pages with at least 100 impressions and compared content types within the same page_1 position tier. Comparison articles had an average CTR of 0.14% (226 pages), while keyword articles had an average CTR of 0.35% (8,186 pages). CTR is stored as a percentage in this dataset, so 0.1412 means 0.14%, not 14%.

Important caveat: the check below shows that all comparison articles in the dataset come from a single client. So this difference may come from that one website rather than from the content type itself.

This does not prove that being a comparison article causes lower CTR. However, it shows an observed difference within a similar ranking-position group, which makes position-adjusted CTR opportunity analysis worth investigating further.

```python
import pandas as pd

# Load the starter dataset from my public repo
url = "https://raw.githubusercontent.com/jjcodes999/flyrank-ml-repo/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Dataset shape:", df.shape)

# Keep pages with enough impressions to make CTR more meaningful
visible = df[df["impressions_90d"] >= 100].copy()

# Compare the two content types within the SAME position tier
page1_comparison = visible[
    (visible["position_tier"] == "page_1") &
    (visible["content_type"].isin([
        "comparison article",
        "keyword article"
    ]))
]

summary = (
    page1_comparison
    .groupby("content_type")
    .agg(
        mean_ctr=("ctr", "mean"),
        page_count=("ctr", "size")
    )
    .round(4)
)

display(summary)
```

```python
# Check: how many different clients (websites) do the comparison articles come from?
comparison = df[df["content_type"] == "comparison article"]

print("Number of comparison articles:", len(comparison))
print("Number of different clients they come from:", comparison["client_id"].nunique())
```
